# Изучение данных

```python
import mne
from matplotlib import pyplot as plt
import os

CH_NAMES = ["ECG_I", "ECG_II", "ECG_III", "ECG_aVR", "ECG_aVL", "ECG_aVF"]
```

## Первый испытуемый

Испытуемый СИС, 21 год, рост 179 см, вес 80кг, пульс на старте 92 уд/мин, насыщение крови кислородом 96%, пульс в конце 93 уд/мин, насыщение крови кислородом 97%. После записи 3 минут в покое (запись 10.34) произошел сбой работы прибора. Следующей записью идут 4 минуты нагрузки и 3 минуты начального периода восстановления (запись 10.39). Третьей записью идут последние 4 минуты периода восстановления (запись 10.49). Описание теста: 5 минут отдых сидя без записи ЭКГ, потом регистрация ЭКГ: 3 минуты покоя, 4 минуты нагрузки + 10 минут восстановительного периода (записываем по протоколу первые 3 минуты и последние 4 минуты).

```python
for file in os.listdir("EEGsignalsMarking/Data/01"):
    raw = mne.io.read_raw_edf(f"EEGsignalsMarking/Data/01/{file}", preload=True, verbose=False)
    print(f"Длина записи {file} {raw.times[-1]/60:.2f} минут")
```

```python
raw1_1 = mne.io.read_raw_edf("EEGsignalsMarking/Data/01/2025-03-17_10.34.edf", preload=True, verbose=False)
raw1_2 = mne.io.read_raw_edf("EEGsignalsMarking/Data/01/2025-03-17_10.39.edf", preload=True, verbose=False)
raw1_3 = mne.io.read_raw_edf("EEGsignalsMarking/Data/01/2025-03-17_10.49.edf", preload=True, verbose=False)
```

```python
plt.figure(figsize=(15, 5))
for i in range(6):
    plt.plot(raw1_1.get_data()[i], label=CH_NAMES[i], color=f"C{i}")
plt.legend()
plt.title("Первая запись целиком")
plt.show()
```

```python
plt.figure(figsize=(15, 5))
for i in range(6):
    plt.plot(raw1_1.get_data()[i][0:10000], label=CH_NAMES[i], color=f"C{i}")
plt.legend()
plt.title("Начало первой записи")
plt.show()
```

```python
plt.figure(figsize=(15, 5))
for i in range(6):
    plt.plot(raw1_2.get_data()[i], label=CH_NAMES[i], color=f"C{i}")
plt.legend()
plt.title("Вторая запись целиком")
plt.show()
```

```python
plt.figure(figsize=(15, 5))
for i in range(6):
    plt.plot(raw1_2.get_data()[i][0:10000], label=CH_NAMES[i], color=f"C{i}")
plt.legend()
plt.title("Начало второй записи")
plt.show()
```

```python
plt.figure(figsize=(15, 5))
for i in range(6):
    plt.plot(raw1_3.get_data()[i], label=CH_NAMES[i], color=f"C{i}")
plt.legend()
plt.title("Третья запись целиком")
plt.show()
```

```python
import mne
from matplotlib import pyplot as plt

fig, axs = plt.subplots(2, 1, figsize=(15, 10))

for i in range(6):
    axs[0].plot(raw1_2.get_data()[i][100000:110000], label=CH_NAMES[i], color=f"C{i}")
    axs[1].plot(raw1_3.get_data()[i][100000:110000], label=CH_NAMES[i], color=f"C{i}")
    axs[0].set_title("Середина второй записи")
    axs[1].set_title("Середина третьей записи")
    axs[0].legend()
    axs[1].legend()

plt.show()
```
